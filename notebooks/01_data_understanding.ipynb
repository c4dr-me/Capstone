```python
from pathlib import Path
import zipfile
import json
import pandas as pd
```

```python
if Path.cwd().name == "notebooks":
    PROJECT_ROOT = Path.cwd().parent
else:
    PROJECT_ROOT = Path.cwd()

RAW_DIR = PROJECT_ROOT / "data" / "raw"

zip_files = list(RAW_DIR.glob("*.zip"))

if not zip_files:
    raise FileNotFoundError(f"No ZIP file found inside {RAW_DIR}")

ZIP_PATH = zip_files[0]

print("Project root:", PROJECT_ROOT)
print("Dataset ZIP:", ZIP_PATH)
print("ZIP size:", round(ZIP_PATH.stat().st_size / (1024 ** 2), 2), "MB")
```

```python
with zipfile.ZipFile(ZIP_PATH, "r") as zip_file:
    files_in_zip = zip_file.namelist()

print("Files inside ZIP:\n")

for file_name in files_in_zip:
    print(file_name)
```

```python
def find_file(filename):
    matches = [
        path
        for path in files_in_zip
        if Path(path).name.lower() == filename.lower()
    ]

    return matches[0] if matches else None


dataset_files = {
    "transactions": find_file("transactions_data.csv"),
    "cards": find_file("cards_data.csv"),
    "users": find_file("users_data.csv"),
    "mcc_codes": find_file("mcc_codes.json"),
    "fraud_labels": find_file("train_fraud_labels.json"),
}

dataset_files
```

```python
with zipfile.ZipFile(ZIP_PATH, "r") as zip_file:
    with zip_file.open(dataset_files["transactions"]) as file:
        transactions_100k = pd.read_csv(
            file,
            nrows=100_000,
            low_memory=False
        )

print("Sample shape:", transactions_100k.shape)
transactions_100k.info()
```

```python
missing_values = (
    transactions_100k
    .isna()
    .sum()
    .sort_values(ascending=False)
)

missing_values

missing_percentages = (
    transactions_100k
    .isna()
    .mean()
    .mul(100)
    .round(2)
    .sort_values(ascending=False)
)

missing_percentages

transactions_100k["errors"].value_counts(dropna=False).head(20)

error_sample = transactions_100k[
    transactions_100k["errors"].notna()
]

print("Error records in sample:", len(error_sample))

display(error_sample.head(20))
```

```python
chunk_size = 500_000

total_transactions = 0
transactions_with_errors = 0
error_counts = pd.Series(dtype="int64")

with zipfile.ZipFile(ZIP_PATH, "r") as zip_file:
    with zip_file.open(dataset_files["transactions"]) as file:

        for chunk_number, chunk in enumerate(
            pd.read_csv(
                file,
                usecols=["errors"],
                chunksize=chunk_size,
                low_memory=False
            ),
            start=1
        ):
            total_transactions += len(chunk)

            errors = (
                chunk["errors"]
                .dropna()
                .astype(str)
                .str.strip()
            )

            errors = errors[errors != ""]

            transactions_with_errors += len(errors)

            chunk_counts = errors.value_counts()

            error_counts = error_counts.add(
                chunk_counts,
                fill_value=0
            )

            print(
                f"Chunk {chunk_number}: "
                f"{total_transactions:,} rows processed"
            )

error_counts = (
    error_counts
    .astype(int)
    .sort_values(ascending=False)
)
```

```python
print("Total transactions:", f"{total_transactions:,}")
print("Transactions with errors:", f"{transactions_with_errors:,}")

error_rate = (
    transactions_with_errors / total_transactions * 100
)

print("Error rate:", round(error_rate, 2), "%")
error_counts
```

```python
individual_error_counts = (
    error_counts
    .rename_axis("combined_error")
    .reset_index(name="count")
    .assign(
        error=lambda df: df["combined_error"].str.split(",")
    )
    .explode("error")
)

individual_error_counts["error"] = (
    individual_error_counts["error"]
    .str.strip()
)

individual_error_counts = (
    individual_error_counts
    .groupby("error")["count"]
    .sum()
    .sort_values(ascending=False)
)

individual_error_counts
```

```python
with zipfile.ZipFile(ZIP_PATH, "r") as zip_file:
    with zip_file.open(dataset_files["cards"]) as file:
        cards = pd.read_csv(file)

    with zip_file.open(dataset_files["users"]) as file:
        users = pd.read_csv(file)

print("Cards shape:", cards.shape)
print("Users shape:", users.shape)
```

```python
card_match_rate = (
    transactions_100k["card_id"]
    .isin(cards["id"])
    .mean()
    * 100
)

user_match_rate = (
    transactions_100k["client_id"]
    .isin(users["id"])
    .mean()
    * 100
)

card_user_match_rate = (
    cards["client_id"]
    .isin(users["id"])
    .mean()
    * 100
)

print(f"Transaction → Card match: {card_match_rate:.2f}%")
print(f"Transaction → User match: {user_match_rate:.2f}%")
print(f"Card → User match: {card_user_match_rate:.2f}%")
```

```python
safe_cards = cards[
    [
        "id",
        "card_brand",
        "card_type",
        "has_chip",
        "credit_limit",
        "card_on_dark_web",
    ]
].rename(columns={"id": "card_id"})

safe_users = users[
    [
        "id",
        "current_age",
        "yearly_income",
        "total_debt",
        "credit_score",
        "num_credit_cards",
    ]
].rename(columns={"id": "client_id"})

joined_sample = (
    transactions_100k
    .merge(safe_cards, on="card_id", how="left")
    .merge(safe_users, on="client_id", how="left")
)

joined_sample["merchant_category"] = (
    joined_sample["mcc"]
    .astype(str)
    .map(mcc_codes)
)

display(
    joined_sample[
        [
            "id",
            "date",
            "client_id",
            "card_id",
            "amount",
            "use_chip",
            "merchant_category",
            "errors",
            "card_brand",
            "card_type",
            "has_chip",
            "credit_limit",
            "credit_score",
            "total_debt",
        ]
    ].head(10)
)
```

```python
fraud_series = pd.Series(fraud_labels["target"], name="fraud_label")

print("Fraud label distribution:")
print(fraud_series.value_counts())

labelled_transaction_ids = set(
    map(int, fraud_labels["target"].keys())
)

sample_label_coverage = (
    transactions_100k["id"]
    .isin(labelled_transaction_ids)
    .mean()
    * 100
)

print(
    f"\nFraud-label coverage in the 100k sample: "
    f"{sample_label_coverage:.2f}%"
)
```
